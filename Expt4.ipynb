```python
# len() IN ARRAY
nums=[10,20,30]
print(len(nums))
nums[2]=50
print(nums)
```

```python
#append
fruits=['mango','kiwi','fig']
fruits.append('apple')
print(fruits)
```

```python
#remove
fruits=['mango','kiwi','fig','apple']
fruits.remove('apple')
print(fruits)
```

```python
#index
num=[1,2,3,2,1]
print(num.index(1))
num.count(3)
```

```python
#extend()
nums=[1,2,3]
num=[4,5,6]
nums.extend(num)
print(nums)
#----------
no=[7,8,9]
no.extend([32,43,54])
print(no)
```

```python
#insert()
num=[1,2,3]
num.insert(3,45)
print(num)
num.insert(2,32)
print(num)
```

```python
#SORT()
# for ascending order
nums=[23,45,65,33,32,98]
nums.sort()
print(nums)
# for descending order
nums.sort(reverse=True)
print(nums)

nums.reverse()
print(nums)
```

```python
fruits=['apple','mango','kiwi']
fruits.sort()
print(fruits)
```
