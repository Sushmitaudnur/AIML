```python
# positional parameters

def add_numbers(a, b):  return a + b
def subtract_numbers(a, b):  return a - b
def multiply_numbers(a, b):  return a * b
def divide_numbers(a, b):  return a / b
def power_numbers(a, b):  return a ** b
add = add_numbers(5, 10)
sub = subtract_numbers(10,5)
mul = multiply_numbers(5, 10)
div = divide_numbers(10, 5)
pow = power_numbers(5, 10)
print("Sum:", add)
print("Difference:", sub)
print("Product:", mul)
print("Division:", div)
print("Power:", pow)
```

```python
# default parameters
def greet(name, message="Hello"): return f"{message}, {name}!"
greeting1 = greet("Alice")
greeting2 = greet("Bob", "Good morning")
print(greeting1)
print(greeting2)
```

```python
# keyword parameters
def describe_pet(pet_name,  animal_type="dog"):
 print(f"I have a {animal_type} named {pet_name}.")
describe_pet(pet_name="Buddy", animal_type="cat")
describe_pet(animal_type="hamster", pet_name="Nibbles")
describe_pet(pet_name="Charlie")
```

```python
# variable length parameters
def make_pizza(size,  *toppings):
 print(f"\nMaking a {size}-inch pizza with the following  toppings:")
 for topping in toppings:
  print(f"- {topping}")
make_pizza(12, "pepperoni", "mushrooms", "green peppers")
make_pizza(16, "extra cheese")
```

```python
def calculate(a, b):  sum = a + b
difference = a - b
product = a * b
quotient = a / b
return sum, difference, product, quotient
s, d, p, q = calculate(10, 5)
print(f"Sum: {s}, Difference: {d},  Product: {p}, Quotient: {q}")
```
