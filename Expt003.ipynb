```python
#CONDITIONAL STATEMENT
#if
age=25
if age>=18:
  print("Eligible")
else :
  print(" Not Eligible")
#elif
marks=85
if marks>=90:
  print("Grade A")
elif marks>=75:
    print("Grade B")
elif marks>=60:
    print("Grade C")
else :
     print("Grade D")
# Nested if
age=25
citizen=True
if age>=18 and citizen:
  print("Eligible")
else :
  print("Not Eligible")
```
